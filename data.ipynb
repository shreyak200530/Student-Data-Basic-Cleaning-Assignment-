```python
import re 
with open("students.csv", "r") as f:
    lines = f.readlines()

header = lines[0].strip().split(",")
rows = [line.strip().split(",") for line in lines[1:]]

print("First 5 rows:", rows[:5])
print("Total rows:", len(rows))
print("Total columns:", len(header))

before_rows = len(rows)
unique_rows = []
for row in rows:
    if row not in unique_rows:
        unique_rows.append(row)
rows = unique_rows
after_rows = len(rows)
duplicates_removed = before_rows - after_rows

header = [h.strip().lower().replace(" ", "_") for h in header]
print(rows)
row[1] = row[1].strip().title()


ages = []
for row in rows:
    match = re.search(r"\d+", row[2])
    if match:
        row[2] = int(match.group())
        ages.append(int(match.group()))
    else:
        row[2] = None
median_age = sorted(ages)[len(ages)//2]
for row in rows:
    if row[2] is None:
        row[2] = median_age



def validate_email(email):
    pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
    return bool(re.match(pattern, email))

invalid_emails = 0
for row in rows:
    email = row[3].strip()
    if not validate_email(email):
        row[3] = "unknown@email.com"
        invalid_emails += 1
    else:
        row[3] = email


for row in rows:
    row[4] = row[4].strip().title()


for row in rows:
    try:
        row[5] = float(row[5])
    except:
        row[5] = 0.0


for row in rows:
    fb = row[6].strip()
    if fb.upper() == "NA":
        fb = ""
    row[6] = fb        


print(rows)

#avg
totals = {}
counts = {}

for row in rows:
    course = row[4]
    score = row[5]
    totals[course] = totals.get(course, 0) + score
    counts[course] = counts.get(course, 0) + 1

for course in totals:
    avg = totals[course] / counts[course]
    print(course, ":", round(avg, 2))

#feedback
given = 0
skipped = 0

for row in rows:
    if row[6].strip(): 
        given += 1
    else:
        skipped += 1

print("Feedback given:", given)
print("Feedback skipped:", skipped)


```
